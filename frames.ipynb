```python
import cv2 as cv
import os
import tkinter as tk
from tkinter import filedialog
```

Laptop Webcam

```python
cap = cv.VideoCapture(0)
```

```python
while True:
    ret, frame = cap.read()       #return T\F, frame
    if not ret:
        print("Can't receive frame. Exiting ...")
        break
    cv.imshow('frame', frame)
    if cv.waitKey(1) == ord('q'):
        print("Quit")
        break

cap.release()
cv.destroyAllWindows()
```

Recorded Video

```python
def select_video_file():
    """Opens a file dialog to let the user select a video file from their PC."""
    # Hide the main tkinter window
    root = tk.Tk()
    root.withdraw()
    
    # Bring the file dialog to the front
    root.attributes('-topmost', True)
    
    # Open the file picker dialog
    print("Opening file explorer... Please select a video file.")
    file_path = filedialog.askopenfilename(
        title="Select a Video File",
        filetypes=[("Video Files", "*.mp4 *.avi *.mov *.mkv *.wmv"), ("All Files", "*.*")]
    )
    
    return file_path

def extract_frames(video_path, output_folder):
    if not video_path:
        print("No file selected. Exiting.")
        return

    # Create output directory
    if not os.path.exists(output_folder):
        os.makedirs(output_folder)
        print(f"Created directory: {output_folder}")
    
    cap = cv.VideoCapture(video_path)
    
    if not cap.isOpened():
        print(f"Error: Could not open video file {video_path}")
        return

    total_frames = int(cap.get(cv.CAP_PROP_FRAME_COUNT))
    print(f"Successfully loaded: {os.path.basename(video_path)}")
    print(f"Total frames to extract: {total_frames}")

    frame_count = 0

    while True:
        success, frame = cap.read()
        
        if not success:
            break
        
        # FIXED: Changed 'cv.imwrite' to 'cv2.imwrite'
        frame_filename = os.path.join(output_folder, f"frame_{frame_count:04d}.jpg")
        cv.imwrite(frame_filename, frame) 
        
        if frame_count % 30 == 0:
            print(f"Extracted frame {frame_count}/{total_frames}")
            
        frame_count += 1

    cap.release()
    print(f"\nDone! Extracted {frame_count} frames to '{output_folder}'.")

# --- Run the Runtime Script ---
if __name__ == "__main__":
    # 1. Select the video at runtime via GUI
    chosen_video = select_video_file()
    
    # 2. Extract frames if a file was chosen
    if chosen_video:
        OUTPUT_DIR = "runtime_extracted_frames"
        extract_frames(chosen_video, OUTPUT_DIR)
    else:
        print("Operation cancelled by user.")
```
